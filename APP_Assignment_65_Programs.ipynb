```python
#Q1. Single Number
#Question: Find the element that appears only once in an array. All other elements appear twice.
def single_number(nums):
    result = 0
    for num in nums:
        result ^= num
    return result

nums = [4, 1, 2, 1, 2]
print("Single Number:", single_number(nums))
```

```python
#Q2. Find Palindromes in a Given Range
#Question: Find all palindrome numbers between two given numbers.Program
def find_palindromes(m, n):
    result = []

    for i in range(m, n + 1):
        if str(i) == str(i)[::-1]:
            result.append(i)

    return result

m = 10
n = 50

print(find_palindromes(m, n))
```

```python
#Q3. Regular Expression Matching
#Question: Implement regular expression matching using . and *.
# . matches any character. * matches zero or more occurrences of the preceding character.
def is_match(s, p):
    dp = [[False] * (len(p) + 1)
          for _ in range(len(s) + 1)]

    dp[0][0] = True

    for j in range(2, len(p) + 1):
        if p[j - 1] == '*':
            dp[0][j] = dp[0][j - 2]

    for i in range(1, len(s) + 1):
        for j in range(1, len(p) + 1):

            if p[j - 1] == '.' or p[j - 1] == s[i - 1]:
                dp[i][j] = dp[i - 1][j - 1]

            elif p[j - 1] == '*':
                dp[i][j] = dp[i][j - 2]

                if p[j - 2] == '.' or p[j - 2] == s[i - 1]:
                    dp[i][j] = dp[i][j] or dp[i - 1][j]

    return dp[len(s)][len(p)]

s = "abcd"
p = ".*"

print(is_match(s, p))
```

```python
#Q4. Decode Ways with Pattern Generation
#Question: Find the total number of ways to decode a numeric string and print all possible decoded strings.
def decode(s):
    result = []

    def backtrack(index, path):
        if index == len(s):
            result.append(path)
            return

        if s[index] == '0':
            return

        num = int(s[index])
        backtrack(index + 1, path + chr(num + 64))

        if index + 1 < len(s):
            num = int(s[index:index + 2])

            if 10 <= num <= 26:
                backtrack(index + 2, path + chr(num + 64))

    backtrack(0, "")
    return result

s = "12"
ans = decode(s)

print("Total Ways:", len(ans))
print("Decoded Strings:")
for x in ans:
    print(x)
```

```python
#Q5. Balanced Binary Tree
#Question: Determine whether a binary tree is height-balanced.
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def height(root):
    if root is None:
        return 0

    left = height(root.left)
    right = height(root.right)

    if left == -1 or right == -1:
        return -1

    if abs(left - right) > 1:
        return -1

    return max(left, right) + 1

def is_balanced(root):
    return height(root) != -1

root = Node(3)
root.left = Node(9)
root.right = Node(20)
root.right.left = Node(15)
root.right.right = Node(7)

print(is_balanced(root))
```

```python
#Q6. Reverse Integer
#Question: Reverse the digits of a signed integer. If the result exceeds the 32-bit integer range, return 0.
def reverse_integer(x):
    sign = -1 if x < 0 else 1
    x = abs(x)

    rev = int(str(x)[::-1])
    rev = rev * sign

    if -2**31 <= rev <= 2**31 - 1:
        return rev
    else:
        return 0

x = -123
print(reverse_integer(x))
```

```python
#Q7. Minimize Maximum Consecutive Zeros with Construction
#Question: Generate all binary strings of length L with exactly K ones. Find the arrangement with the minimum maximum consecutive zeros.
from itertools import combinations

def max_zeros(s):
    return max(len(x) for x in s.split('1'))

def solve(L, K):
    strings = []

    for positions in combinations(range(L), K):
        s = ['0'] * L

        for p in positions:
            s[p] = '1'

        strings.append(''.join(s))

    minimum = min(max_zeros(s) for s in strings)

    print("Possible Strings:")
    for s in strings:
        print(s, "-> max zeros =", max_zeros(s))

    print("Optimal Arrangement(s):")
    for s in strings:
        if max_zeros(s) == minimum:
            print(s, end=" ")

    print("\nMinimum Longest Zero Block:", minimum)

L = 4
K = 2
solve(L, K)
```

```python
#Q8. Merge Triplets to Form Target Triplet
def merge_triplets(triplets, target):
    result = [0, 0, 0]

    for t in triplets:
        if all(t[i] <= target[i] for i in range(3)):
            result = [max(result[i], t[i]) for i in range(3)]

    return result == target

triplets = [[2, 5, 3], [1, 7, 5], [2, 3, 4]]
target = [2, 7, 5]

print(merge_triplets(triplets, target))
```

```python
#Q10. Kth Smallest Element in a BST
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def kth_smallest(root, k):
    stack = []

    while True:
        while root:
            stack.append(root)
            root = root.left

        root = stack.pop()
        k -= 1

        if k == 0:
            return root.val

        root = root.right

root = Node(3)
root.left = Node(1)
root.right = Node(4)
root.left.right = Node(2)

print(kth_smallest(root, 2))
```

```python
#Q12. Coin Change
def coin_change(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1

coins = [1, 2, 5]
amount = 11

print("Minimum Coins:", coin_change(coins, amount))
```

```python
#Q13. Cheapest Flights Within K Stops
def cheapest_flight(n, flights, src, dst, k):
    prices = [float('inf')] * n
    prices[src] = 0

    for _ in range(k + 1):
        temp = prices.copy()

        for u, v, cost in flights:
            if prices[u] != float('inf'):
                temp[v] = min(temp[v], prices[u] + cost)

        prices = temp

    return prices[dst] if prices[dst] != float('inf') else -1

flights = [[0, 1, 100], [1, 2, 100], [0, 2, 500]]

print(cheapest_flight(3, flights, 0, 2, 1))
```

```python
#Q17. Merge Intervals
#Question: Merge all overlapping intervals.
def merge(intervals):
    intervals.sort()
    result = []

    for start, end in intervals:
        if not result or result[-1][1] < start:
            result.append([start, end])
        else:
            result[-1][1] = max(result[-1][1], end)

    return result

intervals = [[1,3],[2,6],[8,10],[15,18]]

print(merge(intervals))
```

```python
#Q18. Maximum Product Subarray
#Question: Find the maximum product of a contiguous subarray.
def max_product(nums):
    maximum = minimum = result = nums[0]

    for num in nums[1:]:
        if num < 0:
            maximum, minimum = minimum, maximum

        maximum = max(num, maximum * num)
        minimum = min(num, minimum * num)

        result = max(result, maximum)

    return result

nums = [2, 3, -2, 4]

print(max_product(nums))
```

```python
#Q19. Minimum Team Selection to Cover Required Skills
# Question: Select the smallest team covering all required skills.
from itertools import combinations

def smallest_team(required, candidates):
    for size in range(1, len(candidates) + 1):
        for team in combinations(range(len(candidates)), size):
            skills = set()

            for i in team:
                skills.update(candidates[i])

            if set(required).issubset(skills):
                return list(team)

required = ["Python", "SQL", "ML", "DS"]

candidates = [
    ["Python", "SQL"],
    ["SQL", "ML"],
    ["ML", "DS"],
    ["DS"]
]

print(smallest_team(required, candidates))
```

```python
#Q20. Word Break with All Patterns
#Question: Check whether a string can be segmented into dictionary words and print all valid segmentations.
def word_break(s, wordDict):
    result = []

    def backtrack(start, path):
        if start == len(s):
            result.append(" ".join(path))
            return

        for word in wordDict:
            if s.startswith(word, start):
                backtrack(start + len(word), path + [word])

    backtrack(0, [])

    print("Can Segment:", bool(result))
    print("Segmentations:", result)

s = "catsanddog"
wordDict = ["cat", "cats", "and", "sand", "dog"]

word_break(s, wordDict)
```

```python
#Q24. Closest Pair of Points
#Question: Find the minimum Euclidean distance between any two points.
import math

def closest_pair(points):
    minimum = float('inf')

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            x1, y1 = points[i]
            x2, y2 = points[j]

            distance = math.sqrt(
                (x2 - x1) ** 2 + (y2 - y1) ** 2
            )

            minimum = min(minimum, distance)

    return minimum

points = [(0, 0), (1, 1)]

print(closest_pair(points))
```

```python
#Q25. Count Trailing Zeros in Binary
#Question: Count the number of zeros at the right end of the binary representation.
def trailing_zeros(n):
    if n == 0:
        return 0

    count = 0

    while n % 2 == 0:
        count += 1
        n //= 2

    return count

n = 12

print("Trailing Zeros:", trailing_zeros(n))
```

```python
#Q30. Set Matrix Zeroes
def set_zeroes(matrix):
    rows = set()
    cols = set()

    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if matrix[i][j] == 0:
                rows.add(i)
                cols.add(j)

    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            if i in rows or j in cols:
                matrix[i][j] = 0

    return matrix

matrix = [[1,1,1],[1,0,1],[1,1,1]]

print(set_zeroes(matrix))
```

```python
#Q31. Card Probability
def card_probability(cards):
    total = len(cards)

    ace = sum(card[:-1] == "A" for card in cards)
    face = sum(card[:-1] in ["J", "Q", "K"] for card in cards)
    red = sum(card[-1] in ["H", "D"] for card in cards)
    black = sum(card[-1] in ["C", "S"] for card in cards)

    print("P(Ace) =", ace / total)
    print("P(Face) =", face / total)
    print("P(Red) =", red / total)
    print("P(Black) =", black / total)

    categories = {"Ace": ace, "Face": face, "Red": red}
    print("Most Frequent Category:",
          max(categories, key=categories.get))

cards = ["AH", "10D", "KC", "QS", "2H", "JD", "7C", "KH"]

card_probability(cards)
```

```python
#Q35. Binary Search
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1

arr = [1, 3, 5, 7, 9]

print(binary_search(arr, 7))
```

```python
#Q39. Matrix Chain Multiplication
def matrix_chain(p):
    n = len(p) - 1
    dp = [[0] * n for _ in range(n)]

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k] + dp[k+1][j]
                    + p[i] * p[k+1] * p[j+1]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[0][n-1]

p = [10, 30, 5, 60]

print("Minimum Cost:", matrix_chain(p))
```

```python
#Q40. Validate and Complete Sudoku Board
def solve_sudoku(board):
    def valid(r, c, num):
        for i in range(9):
            if board[r][i] == num or board[i][c] == num:
                return False

        br = (r // 3) * 3
        bc = (c // 3) * 3

        for i in range(br, br + 3):
            for j in range(bc, bc + 3):
                if board[i][j] == num:
                    return False

        return True

    def backtrack():
        for r in range(9):
            for c in range(9):
                if board[r][c] == '.':
                    for num in '123456789':
                        if valid(r, c, num):
                            board[r][c] = num

                            if backtrack():
                                return True

                            board[r][c] = '.'

                    return False

        return True

    return backtrack()

board = [
    list("53..7...."),
    list("6..195..."),
    list(".98....6."),
    list("8...6...3"),
    list("4..8.3..1"),
    list("7...2...6"),
    list(".6....28."),
    list("...419..5"),
    list("....8..79")
]

if solve_sudoku(board):
    print("Completed Sudoku:")
    for row in board:
        print(" ".join(row))
else:
    print("No solution")
```

```python
#Q41. Find All Words in Grid
def find_words(board, words):
    m, n = len(board), len(board[0])
    result = set()

    def dfs(r, c, word, index):
        if index == len(word):
            return True

        if (r < 0 or c < 0 or r >= m or c >= n
                or board[r][c] != word[index]):
            return False

        ch = board[r][c]
        board[r][c] = '#'

        found = (
            dfs(r+1, c, word, index+1) or
            dfs(r-1, c, word, index+1) or
            dfs(r, c+1, word, index+1) or
            dfs(r, c-1, word, index+1)
        )

        board[r][c] = ch
        return found

    for word in words:
        for i in range(m):
            for j in range(n):
                if dfs(i, j, word, 0):
                    result.add(word)

    return list(result)

board = [
    ["o", "a", "t"],
    ["h", "e", "n"],
    ["i", "s", "r"]
]

words = ["oat", "hen", "hat"]

print(find_words(board, words))
```

```python
#Q42. House Robber
def rob(nums):
    prev1 = 0
    prev2 = 0

    for money in nums:
        current = max(prev1, prev2 + money)
        prev2 = prev1
        prev1 = current

    return prev1

nums = [2, 7, 9, 3, 1]

print("Maximum Money:", rob(nums))
```

```python
#Q43. Find Euler Path
from collections import defaultdict

def euler_path(n, edges):
    graph = defaultdict(list)
    degree = [0] * n

    for u, v in edges:
        graph[u].append(v)
        graph[v].append(u)
        degree[u] += 1
        degree[v] += 1

    odd = [i for i in range(n) if degree[i] % 2]

    print("Node Degrees:")
    for i in range(n):
        print(i, "->", degree[i])

    print("Odd Degree Nodes:", odd)

    if len(odd) not in [0, 2]:
        print("Euler Path Exists: False")
        return

    start = odd[0] if odd else next(
        (i for i in range(n) if degree[i]), 0
    )

    print("Euler Path Exists: True")
    print("Start Node:", start)

    stack = [start]
    path = []

    while stack:
        v = stack[-1]

        if graph[v]:
            u = graph[v].pop()
            graph[u].remove(v)
            stack.append(u)
        else:
            path.append(stack.pop())

    path.reverse()
    print("Euler Path:", " -> ".join(map(str, path)))

n = 4
edges = [[0,1], [1,2], [2,0], [1,3]]

euler_path(n, edges)
```

```python
#Q44. Rotate Image 90 Degrees Clockwise
def rotate(matrix):
    n = len(matrix)

    # Transpose matrix
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = (
                matrix[j][i], matrix[i][j]
            )

    # Reverse each row
    for row in matrix:
        row.reverse()

    return matrix

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(rotate(matrix))
```

```python
#Q45. Climbing Stairs with All Patterns
def climbing_stairs(n):
    patterns = []

    def backtrack(remaining, path):
        if remaining == 0:
            patterns.append(path)
            return

        if remaining >= 1:
            backtrack(remaining - 1, path + [1])

        if remaining >= 2:
            backtrack(remaining - 2, path + [2])

    backtrack(n, [])

    print("Total Ways:", len(patterns))
    print("Patterns:")

    for p in patterns:
        print("+".join(map(str, p)))

climbing_stairs(4)
```

```python
#Q48. Palindrome Number Using Decorator
def palindrome_decorator(func):
    def wrapper(num):
        result = func(num)

        if result:
            print(num, "is a Palindrome")
        else:
            print(num, "is Not a Palindrome")

    return wrapper

@palindrome_decorator
def is_palindrome(num):
    return str(num) == str(num)[::-1]

is_palindrome(121)
is_palindrome(1234)
```

```python
#Q52. CSV Data Cleaning and JSON Conversion
import csv
import json
from io import StringIO

raw_data = """Name,Age,Salary,Join Date
Alice,30,95000,2015-03-01
Bob,,60000,2018-07-15
Charlie,35,Unknown,2024-02-01
Diana,40,110000,2021-11-03"""

lines = raw_data.splitlines()
reader = csv.DictReader(lines)

cleaned_records = []

for row in reader:
    row = {k: v.strip() for k, v in row.items()}

    if not row["Age"] or not row["Salary"]:
        continue

    if row["Salary"] == "Unknown":
        continue

    try:
        record = {
            "name": row["Name"],
            "age": int(row["Age"]),
            "salary": float(row["Salary"]),
            "join_date": row["Join Date"]
        }

        cleaned_records.append(record)

    except ValueError:
        continue

with open("cleaned_data.json", "w") as f:
    json.dump(cleaned_records, f, indent=2)

print(cleaned_records)
```

```python
#Q53. Longest Substring Without Repeating Characters
def longest_substring(s):
    visited = [False] * 26
    left = 0
    maximum = 0

    for right in range(len(s)):
        index = ord(s[right]) - ord('a')

        while visited[index]:
            left_index = ord(s[left]) - ord('a')
            visited[left_index] = False
            left += 1

        visited[index] = True

        maximum = max(maximum, right - left + 1)

    return maximum

s = "abcabcbb"

print("Length of the longest unique substring:",
      longest_substring(s))
```

```python
#Q54. Generator Function to Extract Vowels
def getvowels(name):
    vowels = "aeiouAEIOU"

    for ch in name:
        if ch in vowels:
            yield ch

name = "Priyanka"

for vowel in getvowels(name):
    print(vowel)
```

```python
#Q55. User-Defined Iterator for Decimal Values
class NumberIterator:
    def __init__(self, start, stop, step):
        self.start = start
        self.stop = stop
        self.step = step

    def __iter__(self):
        return self

    def __next__(self):
        if self.start <= self.stop:
            value = self.start
            self.start = round(self.start + self.step, 2)
            return value
        else:
            raise StopIteration

obj = NumberIterator(0.0, 1.0, 0.25)

for value in obj:
    print(value)
```

```python
#Q57. Student Performance Analyzer
import random
import datetime
from collections import Counter
from functools import reduce

students = {
    "Tarun": [82, 91, 78],
    "Karthik": [65, 72, 68],
    "Chadrika": [91, 95, 89],
    "Trisha": [76, 81, 79],
    "Atul": [88, 84, 92]
}

averages = {}

for name, marks in students.items():
    averages[name] = sum(marks) / len(marks)

    total = reduce(lambda x, y: x + y, marks)
    print(name, "Total:", total,
          "Average:", averages[name])

print("Highest Average:", max(averages.values()))

print("Random Student:",
      random.choice(list(students.keys())))

print("Current Date:", datetime.date.today())

grades = []

for avg in averages.values():
    if avg >= 90:
        grades.append("A")
    elif avg >= 75:
        grades.append("B")
    elif avg >= 60:
        grades.append("C")
    else:
        grades.append("D")

print("Grade Classification:", Counter(grades))
```

```python
#Q59. Student Marks Management System
class InvalidMarksError(Exception):
    pass

while True:
    try:
        name = input("Enter student name: ")

        assert name.strip() != "", "Name cannot be empty"

        marks = float(input("Enter marks: "))

        if marks < 0 or marks > 100:
            raise InvalidMarksError("Marks must be 0-100")

        with open("student_records.txt", "a") as f:
            f.write(f"{name}: {marks}\n")

        print("Record saved successfully")

    except ValueError:
        print("Enter numeric marks")

    except AssertionError as e:
        print(e)

    except InvalidMarksError as e:
        print(e)

    finally:
        print("Record processing completed")

    choice = input("Enter another student? (yes/no): ")

    if choice.lower() != "yes":
        break
```

```python
#Q60. Insert Sensor Reading into NumPy Array
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)

try:
    position = int(input("Enter position: "))
    element = input("Enter new element: ")

    if not element.isdigit():
        raise ValueError("Only integer values allowed")

    if position < 0 or position > len(arr):
        raise IndexError("Invalid insertion position")

    arr = np.insert(arr, position, int(element))

    print("Updated Array:", arr)

except ValueError as e:
    print("ValueError:", e)

except IndexError as e:
    print("IndexError:", e)

finally:
    print("Insertion operation completed")
```

```python
#Q61. Precious Metal Loan Management
def calculate_interest(metal, purity, amount, tenure):
    rates = {
        "gold": {18: 8, 22: 7, 24: 6},
        "silver": {925: 9, 950: 8, 999: 7},
        "platinum": {900: 10, 950: 9, 999: 8}
    }

    metal = metal.lower()

    if metal not in rates:
        raise ValueError("Invalid metal type")

    if purity not in rates[metal]:
        raise ValueError("Invalid purity")

    rate = rates[metal][purity]
    interest = amount * rate * tenure / 100

    return interest


try:
    result = calculate_interest("gold", 22, 500000, 2)
    print("Interest Amount:", result)

except TypeError:
    print("Missing arguments")

except ValueError as e:
    print(e)

else:
    print("Calculation successful")

finally:
    print("Loan processing completed")
```

```python
#Q62. Rat in a Maze Using Backtracking
def solve_maze(maze):
    n = len(maze)
    solution = [[0] * n for _ in range(n)]

    def safe(x, y):
        return (0 <= x < n and 0 <= y < n
                and maze[x][y] == 1
                and solution[x][y] == 0)

    def backtrack(x, y):
        if x == n-1 and y == n-1:
            solution[x][y] = 1
            return True

        if safe(x, y):
            solution[x][y] = 1

            directions = [
                (1, 0), (-1, 0),
                (0, 1), (0, -1)
            ]

            for dx, dy in directions:
                if backtrack(x+dx, y+dy):
                    return True

            solution[x][y] = 0

        return False

    if backtrack(0, 0):
        print("Solution Path:")
        for row in solution:
            print("".join(map(str, row)))
    else:
        print("No solution exists")


maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]

solve_maze(maze)
```

```python
#Q64. Create Printer Class and Three Objects
class Printer:
    def __init__(self, capacity, color, printer_type):
        self.capacity = capacity
        self.color = color
        self.printer_type = printer_type

    def display(self):
        print("Capacity:", self.capacity)
        print("Color:", self.color)
        print("Type:", self.printer_type)


printer1 = Printer(20, "Color", "Inkjet")
printer2 = Printer(30, "Mono", "Laser")
printer3 = Printer(25, "Color", "Multifunction")

print("Printer 1:")
printer1.display()
print("Identity:", id(printer1))

print("\nPrinter 2:")
printer2.display()
print("Identity:", id(printer2))

print("\nPrinter 3:")
printer3.display()
print("Identity:", id(printer3))
```
